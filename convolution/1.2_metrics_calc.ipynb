```python
def calc_accuracy(y_true, y_pred):
    """
    Calculate accuracy as the ratio of correct predictions to total predictions.

    Parameters:
    y_true (list): List of true labels.
    y_pred (list): List of predicted labels.

    Returns:
    float: Accuracy score.
    """
    if len(y_true) != len(y_pred):
        raise ValueError("Length mismatch between predictions and true labels")
    if len(y_true) == 0:
        raise ValueError("Empty arrays provided")
    correct = sum(1 for a, b in zip(y_true, y_pred) if a == b)
    return correct / len(y_true)

def calc_precision(y_true, y_pred):
    """
    Calculate precision for binary classification.

    Parameters:
    y_true (list): List of true labels.
    y_pred (list): List of predicted labels.

    Returns:
    float: Precision score.
    """
    if len(y_true) != len(y_pred):
        raise ValueError("Length mismatch between predictions and true labels")
    if len(y_true) == 0:
        raise ValueError("Empty arrays provided")
    TP = sum(1 for yt, yp in zip(y_true, y_pred) if yp == 1 and yt == 1)
    FP = sum(1 for yt, yp in zip(y_true, y_pred) if yp == 1 and yt == 0)
    return TP / (TP + FP) if (TP + FP) > 0 else 0.0

def calc_recall(y_true, y_pred):
    """
    Calculate recall for binary classification.

    Parameters:
    y_true (list): List of true labels.
    y_pred (list): List of predicted labels.

    Returns:
    float: Recall score.
    """
    if len(y_true) != len(y_pred):
        raise ValueError("Length mismatch between predictions and true labels")
    if len(y_true) == 0:
        raise ValueError("Empty arrays provided")
    TP = sum(1 for yt, yp in zip(y_true, y_pred) if yp == 1 and yt == 1)
    FN = sum(1 for yt, yp in zip(y_true, y_pred) if yt == 1 and yp == 0)
    return TP / (TP + FN) if (TP + FN) > 0 else 0.0

def calc_f1(y_true, y_pred):
    """
    Calculate the F1 score for binary classification.

    Parameters:
    y_true (list): List of true labels.
    y_pred (list): List of predicted labels.

    Returns:
    float: F1 score.
    """
    if len(y_true) != len(y_pred):
        raise ValueError("Length mismatch between predictions and true labels")
    if len(y_true) == 0:
        raise ValueError("Empty arrays provided")
    precision = calc_precision(y_true, y_pred)
    recall = calc_recall(y_true, y_pred)
    return (2 * precision * recall / (precision + recall)) if (precision + recall) > 0 else 0.0

def print_confusion_matrix(y_true, y_pred):
    """
    Compute and print the confusion matrix for binary classification.

    Parameters:
    y_true (list): List of true labels.
    y_pred (list): List of predicted labels.
    """
    if len(y_true) != len(y_pred):
        raise ValueError("Length mismatch between predictions and true labels")
    if len(y_true) == 0:
        raise ValueError("Empty arrays provided")
    
    classes = [0, 1]
    matrix = {cls: {c: 0 for c in classes} for cls in classes}

    for yt, yp in zip(y_true, y_pred):
        matrix[yt][yp] += 1
    
    # Print the matrix
    print("Confusion Matrix:")
    print("    " + " ".join(f"{c:3}" for c in classes))
    for cls in classes:
        row = " ".join(f"{matrix[cls][c]:3}" for c in classes)
        print(f"{cls:3} {row}")

# Example Usage
y_true_binary = [0, 1, 1, 0, 1]
y_pred_binary = [0, 1, 0, 0, 1]
print("Binary Classification Metrics")
print("Accuracy:", calc_accuracy(y_true_binary, y_pred_binary))
print("Precision:", calc_precision(y_true_binary, y_pred_binary))
print("Recall:", calc_recall(y_true_binary, y_pred_binary))
print("F1 Score:", calc_f1(y_true_binary, y_pred_binary))
print_confusion_matrix(y_true_binary, y_pred_binary)
```
