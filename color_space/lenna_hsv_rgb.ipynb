```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

image = cv2.imread('data/lenna_color.jpg')

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.axis('off')
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

for ax, (channel, color) in zip(axes, enumerate(["Reds", "Greens", "Blues"])):
    ax.imshow(image_rgb[:, :, channel], cmap=color)
    ax.axis('off')

plt.show()
```

```python
image_hsv = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2HSV)

plt.imshow(image_hsv)
plt.axis('off')
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

for ax, (channel, cmap) in zip(axes, enumerate(["hsv", "gray", "gray"])):
    ax.imshow(image_hsv[:, :, channel], cmap=cmap)
    ax.axis('off')

plt.show()
```

```python
image_hsv_to_rgb = cv2.cvtColor(image_hsv, cv2.COLOR_HSV2RGB)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

axes[0].imshow(image_rgb)
axes[0].set_title("Original RGB")
axes[0].axis('off')

axes[1].imshow(image_hsv_to_rgb)
axes[1].set_title("HSV -> RGB")
axes[1].axis('off')

plt.show()

print("Original RGB mean:", image_rgb.mean(axis=(0, 1)))
print("HSV -> RGB mean:", image_hsv_to_rgb.mean(axis=(0, 1)))
print("Difference:", abs(image_rgb.mean(axis=(0, 1)) - image_hsv_to_rgb.mean(axis=(0, 1))))
```
